# WDI

```python
import wbgapi as wb
import requests
import pandas as pd
```

```python
indicators = {
    'EG.EGY.PRIM.PP.KD': 'energy_intensity_gdp',
    'EN.GHG.ALL.PC.CE.AR5': 'ghg_per_capita',
    'EN.GHG.CO2.RT.GDP.KD': 'carbon_intensity_gdp',
    'EG.FEC.RNEW.ZS': 'renewable_share',
    'EG.USE.PCAP.KG.OE': 'energy_use_pc',
    'NY.GDP.PCAP.KD': 'gdp_per_capita',
}

df = wb.data.DataFrame(list(indicators.keys()), labels=True)
df.to_csv('src/wdi_raw.csv')
```

List of countries

```python

url = "https://api.worldbank.org/v2/country"
params = {"format": "json", "per_page": 400}

response = requests.get(url, params=params)
data = response.json()

countries = pd.json_normalize(data[1])

# tylko realne kraje, bez agregatów regionalnych
countries = countries[countries['incomeLevel.id'] != "NA"]

# najbardziej przydatne kolumny do dalszej pracy
countries = countries[['id', 'iso2Code', 'name', 'region.value', 'incomeLevel.value']]

print(countries.shape)
countries.to_csv('src/countries.csv')
```
